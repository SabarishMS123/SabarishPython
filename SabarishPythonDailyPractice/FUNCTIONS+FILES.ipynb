```python
# Positional Arguments

def student(name, age):
    print("Name :", name)
    print("Age  :", age)

# Function Call
student("Sri Dhivya", 20)
```

```python
# Default Arguments

def greet(name, message="Welcome"):
    print(message, name)

# Function Calls
greet("Sabarish")
greet("Rahul", "Good Morning")
```

```python
# Keyword Arguments

def employee(name, salary):
    print("Name   :", name)
    print("Salary :", salary)

# Function Call
employee(salary=50000, name="Sabarish")
```

```python
# Variable Arguments

def numbers(*values):
    print("Numbers are:")
    for i in values:
        print(i)

# Function Call
numbers(10, 20, 30, 40, 50)
```

```python
# Arbitrary Keyword Arguments

def student_details(**details):
    for key, value in details.items():
        print(key, ":", value)

# Function Call
student_details(name="Sabarish", age=20, department="CSE")
```

```python
# Lambda Function

square = lambda x: x * x

num = int(input("Enter a number: "))
print("Square =", square(num))
```

```python
# Return Function

def add(a, b):
    return a + b

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

result = add(num1, num2)
print("Sum =", result)
```

```python
# Recursive Function

def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)

num = int(input("Enter a number: "))
print("Factorial =", factorial(num))
```
