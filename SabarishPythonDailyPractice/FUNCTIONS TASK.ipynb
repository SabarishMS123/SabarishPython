```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)


n = int(input("Enter a number to find factorial: "))

print("Factorial of", n, "=", factorial(n))
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


n = int(input("Enter a number to check whether it is prime: "))

if is_prime(n):
    print(n, "is a prime number")
else:
    print(n, "is not a prime number")
```

```python
def max_of_three(a, b, c):
    maximum = a

    if b > maximum:
        maximum = b

    if c > maximum:
        maximum = c

    return maximum


a = int(input("Enter number 1: "))
b = int(input("Enter number 2: "))
c = int(input("Enter number 3: "))

print("Maximum of 3 numbers is:", max_of_three(a, b, c))
```

```python
def is_palindrome(string):
    string = string.lower()
    return string == string[::-1]


string = input("Enter a string: ")

if is_palindrome(string):
    print(string, "is a palindrome")
else:
    print(string, "is not a palindrome")
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)


n = int(input("Enter the position of Fibonacci number: "))

print("Fibonacci number =", fibonacci(n))
```

```python
def count_vowels(string):
    count = 0

    for ch in string.lower():
        if ch in "aeiou":
            count += 1

    return count


string = input("Enter a string: ")

print(string, "has", count_vowels(string), "vowels")
```

```python
def reverse_string(string):
    return string[::-1]


string = input("Enter a string: ")

print(string, "reverse is", reverse_string(string))
```

```python
def remove_duplicates(my_list):
    new_list = []

    for item in my_list:
        if item not in new_list:
            new_list.append(item)

    return new_list


my_list = [1, 2, 2, 3, 3, 4, 5, 5]

print("Original list:", my_list)
print("After removing duplicates:", remove_duplicates(my_list))
```

```python
def longest_word(words):
    longest = ""

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest


words = ["saboo", "python", "programming", "AI"]

print("List:", words)
print("Longest word:", longest_word(words))
```

```python
def sum_of_even(numbers):
    total = 0

    for number in numbers:
        if number % 2 == 0:
            total += number

    return total


numbers = [1, 2, 3, 4, 5, 6]

print("List:", numbers)
print("Sum of even numbers:", sum_of_even(numbers))
```

```python
def gcd(a, b):
    if b == 0:
        return a
    else:
        return gcd(b, a % b)


a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("GCD of", a, "and", b, "=", gcd(a, b))
```

```python
def number_of_digits(n):
    return len(str(abs(n)))


n = int(input("Enter a number: "))

print("Number of digits in", n, "=", number_of_digits(n))
```

```python
def sum_of_digits(n):
    total = 0

    for digit in str(abs(n)):
        total += int(digit)

    return total


n = int(input("Enter a number: "))

print("Sum of digits in", n, "=", sum_of_digits(n))
```

```python
def flatten(nested_list):
    flat_list = []

    for item in nested_list:
        if isinstance(item, list):
            for sub_item in item:
                flat_list.append(sub_item)
        else:
            flat_list.append(item)

    return flat_list


nested_list = [[1, 2], [3, 4], 5]

print("Original list:", nested_list)
print("Flattened list:", flatten(nested_list))
```
