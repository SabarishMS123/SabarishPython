```python
#Task 2 :Python program to find the area of circle, square, triangle
import math

print("--- Area Calculator ---")
print("1. Circle")
print("2. Square")
print("3. Triangle")

choice = input("Choose a shape (1-3): ")

if choice == '1':
    radius = float(input("Enter the radius: "))
    area = math.pi * radius * radius
    print("The area is:", round(area, 2))
    
elif choice == '2':
    side = float(input("Enter the side length: "))
    area = side * side
    print("The area is:", round(area, 2))
    
elif choice == '3':
    base = float(input("Enter the base: "))
    height = float(input("Enter the height: "))
    area = 0.5 * base * height
    print("The area is:", round(area, 2))
    
else:
    print("Invalid choice!")
```

```python
#Task 6
print("--- Compound Interest Calculator ---")

money = float(input("How much money are you starting with? "))
rate = float(input("What is the interest rate (e.g., 5 for 5%)? "))
years = float(input("How many years will you save/borrow this? "))

rate_fraction = rate / 100

final_amount = money * ((1 + rate_fraction) ** years)
profit = final_amount - money

print("Your final total will be:", round(final_amount, 2))
print("The interest money you made is:", round(profit, 2))
```

```python
num = int(input("Enter a number: "))
original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = (reverse * 10) + digit
    num = num // 10

if original == reverse:
    print("It is a Palindrome")
else:
    print("It is NOT a Palindrome")
```

```python
num = int(input("Enter a number: "))
original = num
total = 0
power = len(str(num))

while num > 0:
    digit = num % 10
    total = total + (digit ** power)
    num = num // 10

if original == total:
    print("It is an Armstrong number")
else:
    print("It is NOT an Armstrong number")
```

```python
num = int(input("Enter a number: "))
is_prime = True

if num < 2:
    is_prime = False
else:
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

if is_prime:
    print("It is a Prime number")
else:
    print("It is NOT a Prime number")
```

```python
count = 0

for num in range(2, 101):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
        count = count + 1

print("Total prime numbers found:", count)
```
